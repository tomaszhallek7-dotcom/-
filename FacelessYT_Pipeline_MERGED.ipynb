# 🎬 Faceless YouTube Channel — Full Automation Pipeline

This notebook is your **end-to-end production system** for a faceless YouTube channel.
It combines AI script generation, voiceover, stock footage, editing, SEO, and upload —
all with free or low-cost tools.

---

## 📐 Pipeline Overview

```
[1] SETUP          → Define niche, audience, content pillars
[2] SCRIPT         → Claude API generates full narrator script
[3] VOICEOVER      → ElevenLabs converts script to audio
[4] VISUALS        → Stock footage + text animations (CapCut / DaVinci)
[5] EDITING        → Assemble video, add music and captions
[6] SEO PACKAGE    → Claude generates titles, description, tags, chapters
[7] THUMBNAIL      → Claude gives brief → design in Canva
[8] UPLOAD         → YouTube Studio with full metadata
[9] MONITOR        → Track CTR, watch time, iterate
```

**Time per video (target):** ~3-4 hours once your templates are set up.
**Cost:** Can be $0/month using the free tiers listed below.

## 🤖 Step 1 — Claude API Helper

This helper function sends any prompt to Claude and returns the response.
All generation steps below use this function.


````python
import anthropic
from google.colab import userdata
from google.colab.userdata import SecretNotFoundError

ANTHROPIC_API_KEY = None
client = None
ask_claude = None # Initialize to None

try:
    # Fetch the API key from Colab secrets
    ANTHROPIC_API_KEY = userdata.get("ANTHROPIC_API_KEY")
    if not ANTHROPIC_API_KEY: # Treat empty string as not existing
        raise SecretNotFoundError("ANTHROPIC_API_KEY")

except SecretNotFoundError:
    print("❌ ANTHROPIC_API_KEY is not found in Colab secrets or is empty.\n")
    print("To fix this, please follow these steps:")
    print("1. Click on the '🔑' icon (Secrets) in the left sidebar of Colab.")
    print("2. Add a new secret with the name 'ANTHROPIC_API_KEY' and paste your Anthropic API key as its value.")
    print("3. Make sure 'Notebook access' is enabled for this secret.")
    print("After setting the secret, please re-run this cell.")

except Exception as e:
    print(f"An unexpected error occurred while fetching ANTHROPIC_API_KEY: {e}")

else:
    # Only proceed if ANTHROPIC_API_KEY was successfully retrieved
    client = anthropic.Anthropic(api_key=ANTHROPIC_API_KEY)

    def ask_claude(system_prompt: str, user_prompt: str, max_tokens: int = 2000) -> str:
        """Send a prompt to Claude and return the text response."""
        message = client.messages.create(
            model="claude-3-5-sonnet-20240620",
            max_tokens=max_tokens,
            system=system_prompt,
            messages=[{"role": "user", "content": user_prompt}]
        )
        return message.content[0].text

    # Quick test
    try:
        print("\nAttempting a quick test of the Claude API...")
        result = ask_claude(
            "You are a helpful assistant.",
            "Reply with exactly: Claude API is working."
        )
        print("\n✅ Claude API is working! Test result:")
        print(result)
    except Exception as e:
        print(f"\n❌ Failed to run Claude API quick test, even with key set: {e}")
        print("Please double-check your API key or network connection.")
````

**Free tier:** 10,000 characters/month (~7-10 min of audio)  
**Voices:** Browse at https://elevenlabs.io/voice-library

- `AZnzlk1XvdvUeBnXmlld` — Domi (energetic, female)  
- `ErXwobaYiN019PkySvjV` — Antoni (calm, male)


````python
def generate_voiceover(text: str, voice_id: str, output_path: str = "voiceover.mp3") -> str:
    """Generate voiceover using ElevenLabs API."""
    url = f"https://api.elevenlabs.io/v1/text-to-speech/{voice_id}"
    headers = {
        "Accept": "audio/mpeg",
        "Content-Type": "application/json",
        "xi-api-key": ELEVENLABS_API_KEY
    }
    payload = {
        "text": text,
        "model_id": "eleven_multilingual_v2",  # supports Polish, Spanish, French, etc.
        "voice_settings": {
            "stability": 0.5,
            "similarity_boost": 0.75,
            "style": 0.3,
            "use_speaker_boost": True
        }
    }
    response = requests.post(url, json=payload, headers=headers)
    if response.status_code == 200:
        with open(output_path, "wb") as f:
            f.write(response.content)
        print(f"✅ Voiceover saved: {output_path}")
        return output_path
    else:
        print(f"❌ ElevenLabs error {response.status_code}: {response.text}")
        return None

# Strip visual cues from script before sending to TTS
import re
clean_script = re.sub(r"\[VISUAL:[^\]]+\]", "", FULL_SCRIPT).strip()
clean_script = re.sub(r"---+", "", clean_script).strip()

# Uncomment to generate (uses API credits):
# voiceover_file = generate_voiceover(clean_script, VOICE_ID)

print(f"Script length: {len(clean_script)} characters")
print("(Uncomment the last line to generate the voiceover MP3)")
````

### CapCut Workflow:
1. Import voiceover MP3 → it becomes your timeline backbone
2. Add stock clips on track above — match the [VISUAL] cues from the script
3. Use **Auto Captions** (Text → Auto Captions) — adds subtitles automatically
4. Add background music on a separate track at -20dB vs voiceover
5. Add text overlays for key stats/points using the template library
6. Export: 1080p, 30fps, H.264, Target bitrate 8-12 Mbps

### Captions best practices:
- Font: Bold, white with black stroke, centered bottom
- Size: Large enough to read on mobile (subtitle style)
- Color-highlight 1-2 key words per sentence for retention

````python
BATCH_TOPICS = [
    # Add your topics here:
    "5 nawyków porannych najlepszych liderów świata",
    "Jak przestać odkładać wszystko na później (naukowo)",
    "3 błędy, które niszczą Twoją produktywność każdego ranka",
]

batch_results = {}

for i, topic in enumerate(BATCH_TOPICS, 1):
    print(f"\n[{i}/{len(BATCH_TOPICS)}] Processing: {topic}")

    script = ask_claude(
        "You are a YouTube scriptwriter. Write a complete script with [VISUAL:] cues.",
        f"Write a {VIDEO_LENGTH} YouTube script in {CHANNEL_LANGUAGE} for: {topic}",
        max_tokens=2500
    )
    seo = ask_claude(
        "You are a YouTube SEO expert.",
        f"Write SEO package (titles, description, 25 tags) for: {topic} | Niche: {CHANNEL_NICHE} | Language: {CHANNEL_LANGUAGE}",
        max_tokens=1000
    )
    thumb = ask_claude(
        "You are a thumbnail designer.",
        f"Give 2 thumbnail concepts with hex colors and Canva steps for: {topic}",
        max_tokens=800
    )

    batch_results[topic] = {"script": script, "seo": seo, "thumbnail": thumb}

    safe_name = topic[:40].replace(" ", "_").replace("/", "-")
    with open(f"video_{i}_{safe_name}.txt", "w", encoding="utf-8") as f:
        f.write(f"TOPIC: {topic}\n\n")
        f.write("=== SCRIPT ===\n\n" + script + "\n\n")
        f.write("=== SEO ===\n\n" + seo + "\n\n")
        f.write("=== THUMBNAIL ===\n\n" + thumb)

    print(f"   ✅ Saved video_{i}_{safe_name}.txt")

print(f"\n✅ Batch complete — {len(BATCH_TOPICS)} videos generated.")
````

````python
import anthropic

# Initialize the Anthropic client with your API key
client = anthropic.Anthropic(api_key=ANTHROPIC_API_KEY)

def ask_claude(system_prompt: str, user_prompt: str, max_tokens: int = 2000) -> str:
    """Send a prompt to Claude and return the text response."""
    try:
        message = client.messages.create(
            model="claude-3-5-sonnet-20240620", # You can choose other models like 'claude-3-opus-20240229' or 'claude-3-haiku-20240307'
            max_tokens=max_tokens,
            system=system_prompt,
            messages=[{"role": "user", "content": user_prompt}]
        )
        return message.content[0].text
    except anthropic.APIStatusError as e:
        if e.status_code == 401:
            print(f"❌ Anthropic API Error (401 Unauthorized): Your API key is likely invalid or expired.")
            print(f"Please check your ANTHROPIC_API_KEY in Colab secrets.")
        else:
            print(f"❌ Anthropic API Error (Status {e.status_code}): {e.response}")
        raise # Re-raise the exception after printing for further debugging
    except Exception as e:
        print(f"❌ An unexpected error occurred when calling Anthropic API: {e}")
        raise

# Example usage:
print("Asking Claude a question...")
try:
    response = ask_claude(
        system_prompt="You are a friendly assistant.",
        user_prompt="What is the capital of France?"
    )
    print("\nClaude's response:")
    print(response)
except Exception: # Catch the re-raised exception to prevent notebook from stopping immediately
    print("\nFailed to get a response from Claude due to an API error.")
````

````python
import anthropic
from anthropic import RateLimitError
from google.colab import userdata
from tenacity import retry, stop_after_attempt, wait_exponential, retry_if_exception_type
import requests # For ConnectionError

# Ensure ANTHROPIC_API_KEY is loaded for this cell's scope
ANTHROPIC_API_KEY = userdata.get("ANTHROPIC_API_KEY")

@retry(stop=stop_after_attempt(3), wait=wait_exponential(multiplier=1, min=4, max=10),
       retry=retry_if_exception_type((RateLimitError, requests.exceptions.ConnectionError, requests.exceptions.Timeout)))
def ask_claude_with_error_handling(system_prompt: str, user_prompt: str, max_tokens: int = 2000) -> str:
    """Send a prompt to Claude with comprehensive error handling and retry logic."""
    if not ANTHROPIC_API_KEY:
        raise ValueError("ANTHROPIC_API_KEY not found or is empty. Please set it in Colab secrets.")

    try:
        client = anthropic.Anthropic(api_key=ANTHROPIC_API_KEY)
        message = client.messages.create(
            model="claude-3-5-sonnet-20240620", # Or your preferred model
            max_tokens=max_tokens,
            system=system_prompt,
            messages=[{"role": "user", "content": user_prompt}]
        )
        return message.content[0].text
    except RateLimitError as e:
        print(f"❌ Claude API Rate Limit Error: {e}. Retrying...")
        raise # Re-raise to trigger tenacity retry
    except requests.exceptions.ConnectionError as e:
        print(f"❌ Connection Error: {e}. Retrying...")
        raise # Re-raise to trigger tenacity retry
    except requests.exceptions.Timeout as e:
        print(f"❌ Request Timeout: {e}. Retrying...")
        raise # Re-raise to trigger tenacity retry
    except anthropic.APIStatusError as e:
        if e.status_code == 401:
            print(f"❌ Claude API Error (401 Unauthorized): Your API key is likely invalid or expired.")
            print(f"Please check your ANTHROPIC_API_KEY in Colab secrets.")
        elif e.status_code == 400:
            print(f"❌ Claude API Error (400 Bad Request): Check your prompt or parameters. Details: {e.response.text}")
        else:
            print(f"❌ Claude API Error (Status {e.status_code}): {e.response.text}")
        raise # Re-raise the exception after printing for further debugging
    except Exception as e:
        print(f"❌ An unexpected error occurred when calling Anthropic API: {e}")
        raise

print("✅ `ask_claude_with_error_handling` function defined.")

# --- Quick Test of the function ---
print("\nAttempting a quick test of `ask_claude_with_error_handling`...")
try:
    test_result = ask_claude_with_error_handling(
        system_prompt="You are a helpful assistant.",
        user_prompt="Reply with exactly: Error handling test successful."
    )
    print("\n✅ Test Successful:")
    print(test_result)
except Exception as e:
    print(f"\n❌ Test Failed: {e}")
    print("Please ensure your ANTHROPIC_API_KEY is correctly set in Colab secrets and has notebook access.")
````

````python
print("Starting the full automation pipeline (Steps 1-6)...")

# --- Step 1: Generate Topics and Select Best One ---
TOPIC_PROMPT = f"""
Jesteś strategiem treści dla kanału YouTube bez twarzy.

Nisza: {CHANNEL_NICHE}
Język wideo: {CHANNEL_LANGUAGE}
Styl: {CHANNEL_STYLE}
Docelowa długość wideo: {VIDEO_LENGTH}

Wygeneruj 10 pomysłów na tematy wideo. Dla każdego podaj:
1. Tytuł roboczy (angażujący, po polsku)
2. Główny problem/pytanie widza, które wideo rozwiązuje
3. Szacowany potencjał viralowy (1-10) – **dokładnie w formacie: Potencjał: X/10**
4. Główne słowo kluczowe SEO

Format:
## POMYSŁ [N]
Tytuł: ...
Problem: ...
Potencjał: .../10
SEO keyword: ...
"""

print("🔄 Generuję 10 pomysłów na wideo...")
topic_ideas = call_claude(TOPIC_PROMPT, max_tokens=3000)
print(topic_ideas)
(OUTPUT_DIR / "1_topic_ideas.txt").write_text(topic_ideas, encoding="utf-8")

# --- AUTOMATYCZNY WYBÓR NAJLEPSZEGO TEMATU (na podstawie Potencjał: X/10) ---
lines = topic_ideas.splitlines()
best_topic = None
best_score = -1
current_title = None

for line in lines:
    if "Tytuł:" in line:
        current_title = line.split("Tytuł:")[-1].strip()
    if "Potencjał:" in line and current_title:
        match = re.search(r'(\d+)/10', line)
        if match:
            score = int(match.group(1))
            if score > best_score:
                best_score = score
                best_topic = current_title

SELECTED_TOPIC = best_topic if best_topic else "5 nawyków rannych, które zmienią Twój dzień"
print(f"\n🏆 Automatycznie wybrany temat (potencjał {best_score}/10):\n   {SELECTED_TOPIC}")
print("\n✅ Zapisano do: output/1_topic_ideas.txt")


# --- Step 2: Generate Script ---
SCRIPT_SYSTEM = f"""
Jesteś profesjonalnym scenarzystą dla kanałów YouTube bez twarzy.
Piszesz po polsku. Styl: {CHANNEL_STYLE}.
Skrypt musi być gotowy do nagrania przez lektora — bez didaskaliów, opisów wizualnych
ani instrukcji montażu. Tylko czysty tekst narracji.
Długość: dopasuj do {VIDEO_LENGTH} nagrania (ok. 130–150 słów na minutę).
"""

SCRIPT_PROMPT = f"""
Napisz pełny skrypt YouTube dla wideo na temat:
**{SELECTED_TOPIC}**

Struktura:
1. HOOK (pierwsze 30 sekund) — mocne otwarcie, które zatrzymuje scrollowanie
2. WPROWADZENIE — zapowiedź tego, czego widz się dowie
3. ROZWINIĘCIE — 5 sekcji (po jednym nawyku), każda z wyjaśnieniem i praktyczną radą
4. ZAKOŃCZENIE — podsumowanie + CTA (subskrypcja, komentarz)

Używaj:
- Krótkich zdań (maks. 20 słów)
- Pytań retorycznych do angażowania widza
- Konkretnych przykładów i liczb
- Naturalnych przejść między sekcjami

Oznacz każdą sekcję nagłówkiem: [HOOK], [WPROWADZENIE], [NAWYK 1], ..., [ZAKOŃCZENIE]
"""

print(f"\n🔄 Generuję skrypt dla: '{SELECTED_TOPIC}'...")
script = call_claude(SCRIPT_PROMPT, system=SCRIPT_SYSTEM, max_tokens=6000)
print(script[:800] + "..." if len(script) > 800 else script)

# Zapisz skrypt
script_path = OUTPUT_DIR / "2_script.txt"
script_path.write_text(script, encoding="utf-8")

# Policz słowa i ostrzeż jeśli za długie
word_count = len(script.split())
estimated_minutes = word_count / 140
print(f"\n📊 Statystyki skryptu:")
print(f"   Słów: {word_count}")
print(f"   Szacowany czas nagrania: {estimated_minutes:.1f} min")
if estimated_minutes > 12:
    print(f"   ⚠️ UWAGA: Skrypt jest za długi (docelowo {VIDEO_LENGTH}). Rozważ skrócenie.")
print(f"✅ Zapisano do: output/2_script.txt")


# --- Step 3: Voiceover (ElevenLabs) ---
def text_to_speech_elevenlabs_v3(
    text: str,
    output_path: Path,
    voice_id: str = VOICE_ID,
    chunk_size: int = 4000,
) -> Path:
    """
    Konwertuje tekst na mowę przez ElevenLabs i poprawnie scala fragmenty MP3.
    """
    if not ELEVENLABS_API_KEY:
        raise ValueError("Brak ELEVENLABS_API_KEY — dodaj do Secrets")

    # Dzielenie tekstu na fragmenty (bez przecinania wyrazów)
    chunks = []
    remaining = text
    while len(remaining) > chunk_size:
        split_at = remaining.rfind('. ', 0, chunk_size)
        if split_at == -1:
            split_at = remaining.rfind(' ', 0, chunk_size)
        if split_at == -1:
            split_at = chunk_size
        else:
            split_at += 1  # zachowaj kropkę
        chunks.append(remaining[:split_at])
        remaining = remaining[split_at:].lstrip()
    if remaining:
        chunks.append(remaining)

    print(f"   Fragmenty do wysłania: {len(chunks)}")
    combined = AudioSegment.empty()

    for i, chunk in enumerate(chunks, 1):
        print(f"   🔊 Generuję fragment {i}/{len(chunks)} ({len(chunk)} znaków)...")
        url = f"https://api.elevenlabs.io/v1/text-to-speech/{voice_id}"
        headers = {
            "xi-api-key": ELEVENLABS_API_KEY,
            "Content-Type": "application/json",
        }
        payload = {
            "text": chunk,
            "voice_settings": {
                "stability": 0.5,
                "similarity_boost": 0.75,
            },
        }
        resp = requests.post(url, json=payload, headers=headers)
        if resp.status_code != 200:
            raise RuntimeError(f"ElevenLabs błąd {resp.status_code}: {resp.text}")

        audio_seg = AudioSegment.from_mp3(BytesIO(resp.content))
        combined += audio_seg
        print(f"   ✅ Fragment {i} dodany ({len(audio_seg)/1000:.1f}s)")

    combined.export(output_path, format="mp3")
    size_kb = output_path.stat().st_size / 1024
    print(f"   ✅ Audio zapisane: {output_path} ({size_kb:.0f} KB, {len(combined)/1000:.1f}s)")
    return output_path


# Załaduj skrypt i usuń nagłówki sekcji (zostawiając resztę interpunkcji)
script_text = (OUTPUT_DIR / "2_script.txt").read_text(encoding="utf-8")
clean_script = re.sub(r"\[[^\]]+\]", "", script_text)  # usuń tylko [HOOK], [NAWYK 1] itp.
clean_script = re.sub(r'\n\s*\n', '\n\n', clean_script).strip()

print("\n🔄 Generuję voiceover (poprawne scalanie MP3)...")
audio_path = text_to_speech_elevenlabs_v3(clean_script, OUTPUT_DIR / "3_voiceover.mp3")
print("✅ Voiceover gotowy!")


# --- Step 4: SEO Package ---
SEO_PROMPT = f"""
Jesteś ekspertem SEO dla YouTube z 5-letnim doświadczeniem w polskim rynku.

Temat wideo: {SELECTED_TOPIC}
Nisza: {CHANNEL_NICHE}
Język: {CHANNEL_LANGUAGE}

Wygeneruj pełen pakiet SEO:

### 1. TYTUŁY (5 wariantów, każdy maks. 60 znaków)
Używaj liczb, emocji, ciekawości. Zaznacz rekomendowany gwiazdką (*).

### 2. OPIS
- Pierwsze 2 zdania: hook z głównym słowem kluczowym (wyświetlane bez rozwijania)
- Główna treść (200–250 słów): co widz zyska, punkty z wideo
- Słowa kluczowe organicznie wplecione (co najmniej 8 różnych)
- Sekcja ROZDZIAŁY (timestamps — użyj placeholder [00:00], [01:30] etc.)
- Sekcja LINKI: Media społecznościowe i newsletter (placeholders)
- Sekcja HASHTAGI: 5 tagów

### 3. TAGI
30 tagów SEO oddzielonych przecinkami (mix: krótkie + długie frazy, po polsku i angielsku)

### 4. KATEGORIA I USTAWIENIA
Rekomendowana kategoria YouTube + sugestia dotycząca kart i ekranów końcowych
"""

print("\n🔄 Generuję pakiet SEO...")
seo_package = call_claude(SEO_PROMPT, max_tokens=3000)
print(seo_package[:1000] + "..." if len(seo_package) > 1000 else seo_package)

seo_path = OUTPUT_DIR / "4_seo_package.txt"
seo_path.write_text(seo_package, encoding="utf-8")
print(f"\n✅ Zapisano do: output/4_seo_package.txt")


# --- Step 5: Thumbnail Brief ---
THUMBNAIL_PROMPT = f"""
Jesteś dyrektorem kreatywnym specjalizującym się w miniaturach YouTube.
Wiesz, że miniatura musi działać w rozmiarze 100x56px (na mobile) i 1280x720px.

Temat wideo: {SELECTED_TOPIC}
Nisza: {CHANNEL_NICHE}
Styl kanału: {CHANNEL_STYLE}

Wygeneruj 3 różne koncepcje miniatur. Dla każdej podaj:

#### KONCEPCJA [N] — [Nazwa koncepcji]

**Tekst na miniaturze:**
- Główny nagłówek: (maks. 3 słowa, CAPS)
- Podtytuł: (opcjonalny, maks. 4 słowa)

**Kolorystyka:**
- Tło: [kolor HEX + opis]
- Tekst: [kolor HEX]
- Akcent: [kolor HEX]

**Elementy wizualne:**
- Zdjęcie stock: [konkretne hasło do Pexels/Unsplash]
- Ikony/strzałki: [opis]
- Układ: [opis pozycjonowania elementów]

**Czcionka:** (wybierz z Canvy)

**Dlaczego ta koncepcja zadziała:** (1-2 zdania psychologicznego uzasadnienia)

**Prompt do Canva Magic Design:**
(Gotowy prompt, który można wkleić do Canvy)
"""

print("\n🔄 Generuję brief do miniatury...")
thumbnail_brief = call_claude(THUMBNAIL_PROMPT, max_tokens=2500)
print(thumbnail_brief[:1000] + "..." if len(thumbnail_brief) > 1000 else thumbnail_brief)

thumb_path = OUTPUT_DIR / "5_thumbnail_brief.txt"
thumb_path.write_text(thumbnail_brief, encoding="utf-8")
print(f"\n✅ Zapisano do: output/5_thumbnail_brief.txt")


# --- Step 6: Pre-publication Checklist ---
output_files = list(OUTPUT_DIR.glob("*"))
print("\n📂 Wygenerowane pliki:")
for f in sorted(output_files):
    size = f.stat().st_size
    unit = "KB" if size > 1024 else "B"
    val  = size // 1024 if size > 1024 else size
    print(f"   ✅ {f.name:35s} ({val} {unit})")

print()
print("─" * 60)
print("📋 LISTA KONTROLNA — PRZED PUBLIKACJĄ")
print("─" * 60)
checklist = [
    ("AUTOMATYCZNE — gotowe ✅", [
        "Skrypt narratora (output/2_script.txt)",
        "Voiceover .mp3 (output/3_voiceover.mp3) – sprawdź czy nie ma artefaktów",
        "Pakiet SEO: tytuły, opis, tagi (output/4_seo_package.txt)",
        "Brief do miniatury × 3 koncepcje (output/5_thumbnail_brief.txt)",
    ]),
    ("DO ZROBIENIA RĘCZNIE 🖐️", [
        "[ ] Odsłuchaj voiceover – czy fragmenty są płynnie scalone",
        "[ ] Pobierz footage stock z Pexels.com / Pixabay.com",
        "[ ] Zmontuj wideo w CapCut / DaVinci Resolve (skrypt + audio + footage)",
        "[ ] Dodaj napisy (auto w CapCut lub Whisper)",
        "[ ] Dodaj muzykę tła (YouTube Audio Library — bezpłatna)",
        "[ ] Zaprojektuj miniaturę w Canvie (wg briefu z kroku 5)",
        "[ ] Wyeksportuj wideo: 1080p, H.264, .mp4",
        "[ ] Wgraj do YouTube Studio z pakietem SEO (krok 4)",
        "[ ] Dodaj karty i ekran końcowy (subskrypcja + polecany film)",
        "[ ] Zaplanuj test A/B miniatury (Test & Compare)",
        "[ ] Po 7 dniach: sprawdź CTR w YouTube Analytics",
    ]),
]
for section, items in checklist:
    print(f"\n{section}")
    for item in items:
        print(f"   {item}")

print()
print("─" * 60)
print("🎉 Pipeline v3 zakończony! Powodzenia z kanałem.")
print("─" * 60)
````

````python
from reportlab.lib.pagesizes import letter
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.units import inch

def generate_pdf_summary(markdown_content: str, output_filename: str):
    doc = SimpleDocTemplate(output_filename, pagesize=letter)
    styles = getSampleStyleSheet()

    # Define a style for list items
    styles.add(ParagraphStyle(name='ListItem',
                              parent=styles['Normal'],
                              leftIndent=0.2 * inch,
                              firstLineIndent=-0.2 * inch,
                              bulletIndent=0,
                              bulletFontName='Helvetica',
                              bulletFontSize=10,
                              leading=12))

    story = []

    # Split content by lines to process headings, list items, and paragraphs
    lines = markdown_content.split('\n')
    for line in lines:
        if line.startswith('## '):
            # Handle H2 headings
            story.append(Paragraph(line[3:].strip(), styles['h2']))
            story.append(Spacer(1, 0.2 * inch))
        elif line.startswith('# '):
            # Handle H1 headings
            story.append(Paragraph(line[2:].strip(), styles['h1']))
            story.append(Spacer(1, 0.2 * inch))
        elif line.startswith('* ') or line.startswith('- '):
            # Handle list items
            # The bullet character needs to be explicitly added to the text
            story.append(Paragraph(f'&bull; {line[2:].strip()}', styles['ListItem']))
            story.append(Spacer(1, 0.05 * inch)) # Smaller space for list items
        elif line.strip(): # Avoid empty paragraphs
            # Handle regular paragraphs
            story.append(Paragraph(line.strip(), styles['Normal']))
            story.append(Spacer(1, 0.1 * inch))

    doc.build(story)
    print(f'Generated PDF: {output_filename}')
````

````python
%%capture
!pip uninstall -y optillm transformers accelerate
!pip install transformers==4.23.1
!pip install accelerate
!pip install optillm
!pip install math_verify

from transformers import AutoModelForCausalLM, AutoTokenizer
from optillm.cot_decoding import cot_decode
from optillm.entropy_decoding import entropy_decode
import torch
import time
import pandas as pd # Import pandas

model_name = "facebook/opt-350m" # Changed model name
model = AutoModelForCausalLM.from_pretrained(model_name, output_attentions=True, attn_implementation="eager")
tokenizer = AutoTokenizer.from_pretrained(model_name)

messages = [
    {"role": "user", "content": "A factory produces widgets. In January, it produced 1000 widgets. In February, production increased by 15% compared to January. In March, production decreased by 10% compared to February. If each widget sells for $5, and the cost to produce each widget is $3, what was the total profit for the first quarter (January, February, March)?"}
]

# --- Common setup for all decoding methods --- General setup before decoding
def get_device():
  if torch.cuda.is_available():
    return torch.device("cuda")
  else:
    return torch.device("cpu")

device = get_device()
model = model.to(device)

# Prepare input with proper attention mask
# Using tokenizer.apply_chat_template might not be appropriate for all models, especially older ones like OPT.
# For a general comparison, raw tokenization might be more reliable if chat template is not defined or is causing issues.
# However, sticking to the existing pattern for now.
# Fallback to simple tokenization if chat_template is not set
if tokenizer.chat_template is None:
    print("Warning: tokenizer.chat_template is not set. Falling back to simple tokenization.")
    input_text = messages[0]["content"] # Assuming single user message for simplicity
    input_ids = tokenizer.encode(input_text, return_tensors="pt")
else:
    input_ids = tokenizer.apply_chat_template(messages, tokenize=True, add_generation_prompt=True, return_tensors="pt")

attention_mask = torch.ones_like(input_ids)  # Create attention mask
input_ids = input_ids.to(device)
input_length = input_ids.shape[1]
attention_mask = attention_mask.to(device)

# Get pad and eos token ids for greedy
pad_token_id = tokenizer.pad_token_id
if pad_token_id is None:
    pad_token_id = tokenizer.eos_token_id
# --- End common setup ---

print(f"\n--- Testing Model: {model_name} ---")

# Generate the response using CoT decoding
start_time_cot = time.time()
# Assuming optillm functions handle device transfer or work with model on device.
result_cot, confidence = cot_decode(model, tokenizer, messages, aggregate_paths=True, max_new_tokens=1024)
end_time_cot = time.time()

# Generate the response using Entropy decoding
start_time_entropy = time.time()
result_entropy = entropy_decode(model, tokenizer, messages, max_new_tokens=1024)
end_time_entropy = time.time()

# --- Greedy decoding ---
start_time_greedy = time.time()
output_ids_greedy = model.generate(
    input_ids,
    attention_mask=attention_mask,
    max_new_tokens=1024,
    do_sample=False,     # Greedy decoding
    num_beams=1,        # Single beam for greedy
    pad_token_id=pad_token_id,
    no_repeat_ngram_size=2, # Add this to prevent repetition
    repetition_penalty=1.2, # Add this to penalize repetition
    use_cache=True    # Enable KV caching for faster generation
)
end_time_greedy = time.time()

output_ids_greedy = output_ids_greedy.cpu()
# Decode only the newly generated tokens
response_greedy = tokenizer.decode(output_ids_greedy[0][input_length:], skip_special_tokens=True)

# Consolidate results for table and CSV
def get_fragment(text, length=100):
    # Basic sanitization for markdown table (replace |) and CSV (remove newlines)
    clean_text = text.replace('|', '/').replace('\n', ' ')
    if len(clean_text) > length:
        return clean_text[:length] + '...'
    return clean_text

results_data = []

results_data.append({"Method": "CoT Decoding",
                     "Result_Fragment": get_fragment(result_cot),
                     "Generation_Time_s": f"{end_time_cot - start_time_cot:.2f}",
                     "Full_Result": result_cot})

results_data.append({"Method": "Entropy Decoding",
                     "Result_Fragment": get_fragment(result_entropy),
                     "Generation_Time_s": f"{end_time_entropy - start_time_entropy:.2f}",
                     "Full_Result": result_entropy})

results_data.append({"Method": "Greedy Decoding",
                     "Result_Fragment": get_fragment(response_greedy),
                     "Generation_Time_s": f"{end_time_greedy - start_time_greedy:.2f}",
                     "Full_Result": response_greedy})

# Create DataFrame and save to CSV
df_results = pd.DataFrame(results_data)
csv_filename = "decoding_comparison_results.csv"
df_results.to_csv(csv_filename, index=False)
print(f"\nWyniki porównania zostały zapisane do pliku: {csv_filename}")

# Generate markdown table
markdown_table = "### Porównanie Wyników Dekodowania\n\n"
markdown_table += "Poniższa tabela podsumowuje wyniki i czasy generowania dla każdej z metod dekodowania. Pełne teksty wygenerowanych odpowiedzi znajdują się w wynikach wykonania tej komórki kodu.\n\n"
markdown_table += "| Metoda Dekodowania | Wynik (fragment) | Czas Generowania (s) |\n"
markdown_table += "| :----------------- | :----------------- | :------------------- |\n"
for row in results_data:
    markdown_table += f"| **{row['Method']}** | `{row['Result_Fragment']}` | `{row['Generation_Time_s']}` |\n"

# Print the markdown table
print(markdown_table)

# Print full results for reference (optional, but good for context)
print(f"\n--- Pełne Wyniki Dekodowania ---")
print(f"\nCoT Decoding Result:\n {result_cot}")
print(f"\nEntropy Decoding Result:\n {result_entropy}")
print(f"\nGreedy Decoding Result:\n {response_greedy}")
````
